```python

# ==========================================
# TUGAS DATA SCIENCE - SESI 7
# EKSPLORASI DATASET IRIS
# Nama  : Wirya Cahya Saputra
# NIM   : 20240040315
# ==========================================

# 1. Import library
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Load dataset Iris
df = sns.load_dataset('iris')

# 3. Menampilkan 5 baris pertama
print("5 BARIS PERTAMA DATASET")
print(df.head())

# 4. Informasi dataset
print("\nINFORMASI DATASET")
print(df.info())

# 5. Statistik deskriptif
print("\nSTATISTIK DESKRIPTIF")
print(df.describe())

# 6. Visualisasi distribusi kelas
plt.figure(figsize=(7, 5))
sns.countplot(x='species', data=df)
plt.title('Distribusi Kelas Bunga Iris')
plt.xlabel('Jenis Bunga')
plt.ylabel('Jumlah Data')
plt.show()

# 7. Scatter Plot
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='sepal_length',
    y='petal_length',
    hue='species',
    data=df,
    s=70
)
plt.title('Scatter Plot Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend(title='Species')
plt.show()

# 8. Matriks Korelasi
corr_matrix = df.select_dtypes(include='number').corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title('Matriks Korelasi Dataset Iris')
plt.show()

# 9. Kesimpulan
print("\nKESIMPULAN")
print("Dataset Iris memiliki 150 data dan 3 jenis bunga.")
print("Setiap jenis bunga memiliki 50 data.")
print("Petal length dan petal width memiliki korelasi positif yang kuat.")
print("Scatter plot menunjukkan perbedaan karakteristik antarjenis bunga.")
```
